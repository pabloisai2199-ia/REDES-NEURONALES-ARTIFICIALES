### Calcular producto punto

El primer paso es calcular el valor intermedio, *h*, a partir del producto punto. La fórmula explícita es la siguiente:

$$ h = W X +b $$

```python
# TODO: (1 punto) Implementar la función h sin utilizar funciones predefinidas de numpy como numpy.dot()
#     * Asegúrate de que W y X tengan la misma forma.
#     * Imprime h durante el debug para ver el valor intermedio.

def combinacion_lineal(W, X, b):
    if not isinstance(X[0], (list, tuple)):
        X = [X]

    for fila in X:
        if len(fila) != len(W):
            raise ValueError("W y X no tienen la misma forma")
        else:pass

    resultados = []
    for fila in X:                # recorremos cada muestra
        suma = 0
        for i in range(len(W)):   # recorremos cada entrada
            suma += W[i] * fila[i]
        suma += b
        resultados.append(suma)

    print(f"Valor intermedio h: {resultados}")
    return resultados
```

### Función de activación

Para este ejemplo utilizaremos la función escalón como función de activación.

$$ 	f(h) = \begin{cases}
		0 & \text{if } h < a \\
		1 & \text{if } h \geq a
\end{cases}  $$

```python
# TODO: (1 punto) Completar el código
# Tips:
# - La función escalón debe devolver 0 cuando h < a y 1 cuando h >= a.
# - Puedes elegir un umbral 'a' (por ejemplo 0) o utilizar uno definido fuera de la función.
# - h puede ser un escalar o un array de numpy: usa operaciones vectorizadas para que funcione en ambos casos.
# - Una forma común es usar la comparación (h >= a) y convertir el array booleano a enteros (0/1).

def escalon(h, a=0):
    return (np.asanyarray(h) >= a).astype(int)
```

```python
# Caso escalar
print(escalon(0.5))
print(escalon(-0.3))

# Caso array
h = np.array([-1, 0, 1, 2])
print(escalon(h))

# Con umbral distinto
print(escalon(h, a=1))
```

## Definir perceptrón

Perceptrón como una función

```python
# Perceptrón simple
def perceptron(W, X, b, activacion):
    h = combinacion_lineal(W, X, b)
    return activacion(h)
```

```python
import numpy as np

def combinacion_lineal(W, X, b):
    if not isinstance(X[0], (list, tuple)):
        X = [X]

    for fila in X:
        if len(fila) != len(W):
            raise ValueError("W y X no tienen la misma forma")
        else:pass

    resultados = []
    for fila in X:                # recorremos cada muestra
        suma = 0
        for i in range(len(W)):   # recorremos cada entrada
            suma += W[i] * fila[i]
        suma += b
        resultados.append(suma)

    print(f"Valor intermedio h: {resultados}")
    return resultados

def escalon(h, a=0):
    return (np.asanyarray(h) >= a).astype(int)

def perceptron(W, X, b, activacion):
    h = combinacion_lineal(W, X, b)
    return activacion(h)

inputs_or = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])
outputs_esperados = np.array([0,1,1,1])

weights = np.array([0.5,0.5])
bias = 0.1


print(f"Pesos: {weights}, Sesgo: {bias}")
print("-" * 40)


correctos = 0  # Inicializar contador

for i in range(len(inputs_or)):
    output = perceptron(weights, inputs_or[i], bias, escalon)
    esperado = outputs_esperados[i]
    acierto = "✓" if output == esperado else "✗"
    print(f"OR({inputs_or[i][0]}, {inputs_or[i][1]}) = {output}  (esperado: {esperado})  {acierto}")
    if output == esperado:
        correctos += 1

print("-" * 40)
print(f"Precisión: {correctos}/{len(inputs_or)} = {correctos/len(inputs_or)*100}%")
print("-" * 40)


combinaciones = [
    ([1.0, 1.0], -0.5),
    ([0.5, 0.5], -0.1),
    ([2.0, 2.0], -1.0),
    ([0.8, 0.3], -0.2),
    ([0.6, 0.6], -0.3),
]

for idx, (weights, bias) in enumerate(combinaciones, 1):
    print(f"Prueba {idx}: Pesos = {weights}, Sesgo = {bias}")
    print("-" * 50)

    correctos = 0
    for i in range(len(inputs_or)):
        output = perceptron(weights, inputs_or[i], bias, escalon)
        esperado = outputs_esperados[i]
        acierto = "✓" if output == esperado else "✗"
        print(f"OR({inputs_or[i][0]}, {inputs_or[i][1]}) = {output}  (esperado: {esperado})  {acierto}")
        if output == esperado:
            correctos += 1

    print(f"Precisión: {correctos}/4 = {correctos/4*100}%")
    if correctos == 4:
        print("🎉 ¡ENCONTRADO! Estos pesos funcionan para OR")
    print("\n")
```
